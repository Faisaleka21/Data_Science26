```python
semester_cols = ['ip1', 'ip2', 'ip3', 'ip4']
semester_group = df.groupby('graduation')[semester_cols].mean()
semester_group
```

```python
semester_plot = semester_group.T

semester_plot.index = [
    'Semester 1',
    'Semester 2',
    'Semester 3',
    'Semester 4'
]

semester_plot
```

```python
semester_plot.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Rata-rata GPA Semester 1–4 berdasarkan Status Kelulusan')
plt.xlabel('Semester')
plt.ylabel('Rata-rata GPA')
plt.xticks(rotation=0)
plt.legend(title='Status Kelulusan')
plt.show()
```
