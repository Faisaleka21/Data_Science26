```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y='Class',
    order=df['Class'].value_counts().index
)

plt.title('Distribusi Tingkat Obesitas')
plt.xlabel('Jumlah')
plt.ylabel('Class')
plt.show()
```

Dari grafik tersebut yang saya temukan yaitu dari tingkatan obsesitas jumlah data paling banyak berada pada kategori Obesity_Type_I dengan jumlah sekitar 350, sedangkan yang paling sedikit adalah kategori Insufficient_Weight sekitar 270 data.

```python
df.groupby('Class')['Weight'].mean().sort_values()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Weight',
    y='Class'
)

plt.title('Distribusi Weight berdasarkan Tingkat Obesitas')
plt.xlabel('Weight')
plt.ylabel('Tingkat Obesitas')
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Height',
    y='Weight'
)

plt.title('Hubungan Height dan Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Height',
    y='Weight',
    hue='Class'
)

plt.title('Height dan Weight berdasarkan Tingkat Obesitas')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.legend(
    title='Tingkat Obesitas',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)
plt.show()
```
