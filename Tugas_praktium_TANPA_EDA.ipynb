```python
# Program 2, Tanpa EDA
import pandas as pd
from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# 1. Baca dataset
df = pd.read_csv('https://raw.githubusercontent.com/Faisaleka21/Data_Science26/refs/heads/main/Data_Set/train.csv')

df = df[["Pclass", "Sex", "Age", "Survived"]]
```

```python
# 2. Bagi data
train, test = train_test_split(
df, test_size=0.2,
random_state=42,
stratify=df["Survived"]
)
```

```python
# 3. Preprocessing langsung
def proses(data):
  data = data.copy()
  data["Age"] = data["Age"].fillna(
      train["Age"].median()
      )
  data["Sex"] = data["Sex"].map({
      "male": 0, "female": 1
      })
  return data

train = proses(train)
test = proses(test)

fitur = ["Pclass","Sex","Age"]
```

```python
# 4. Training
model = DecisionTreeClassifier(
    max_depth=3, random_state=42
)
model.fit(train[fitur], train["Survived"])
```

```python
# 5. Evaluasi
prediksi = model.predict(test[fitur])
print("Akurasi tanpa EDA:",
accuracy_score(test["Survived"], prediksi))
```
